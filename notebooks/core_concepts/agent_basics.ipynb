## Creating a Simple Agent

#### Function: create_agent

create_agent is a prebuilt function for creating and invoking an agent graph. At initialization it requires the model argument to select the LLM to power the agent.

```python
agent = create_agent(model='claude-sonnet-5')
```

Invoking the agent requires passing a dictionary of messages. Use the HumanMessage class from the messages module to tell the model the message is from the user.

```python
response = agent.invoke({'messages':[HumanMessage(content="Who was Ernest Shackleton?")]})
```

The reponse has the actual response from the LLM and other metadata. Access the content of the last returned message to get the tokens from the model

```python
pprint(response['messages'][-1].content)
```
