## Configuring the MCP client

langhchain_mcp_adapters library has a class called MultiServerMCPClient that is used to connect to the MCP server. It accepts a dict arugment listing the details of the server; in this case a local, custom MCP server defined in the same directory written in python.

Note: transport can either be standard IO or streamable HTTP. Documentation on transports can be found here: https://modelcontextprotocol.io/specification/2026-07-28/basic/transports.

```python
client = MultiServerMCPClient(
    {
        'local_server': {
            'transport': 'stdio',
            'command': 'python',
            'args': ['src/custom_mcp_server.py']
        }
    }
)
```

```python
# Getting tools from the client
tools = await client.get_tools()

# Getting resources from the client
resources = await client.get_resources('local_server')

# Get prompts
prompt = await client.get_prompt('local_server', 'prompt')
useful_prompt = prompt[0].content
```

```python
agent = create_agent(
    model='claude-sonnet-5',
    tools=tools,
    system_prompt=useful_prompt
)
```

```python
response = await agent.ainvoke(
    {'messages': [HumanMessage(content="Tell me about the langchain-mcp-adapters library")]}
    )
```

```python
pprint(response)
```
